```python
import pandas as pd
import numpy as np
```

```python
df_gmc = pd.read_csv('/content/drive/MyDrive/credit_score_project/datasets/GiveMeSomeCredit/cs-training.csv')
```

```python
df_gmc.drop(columns=['Unnamed: 0'],inplace=True)

renamer = {'SeriousDlqin2yrs':'defaults',
'RevolvingUtilizationOfUnsecuredLines':'RUUL',
'NumberOfTime30-59DaysPastDueNotWorse':'30-59 passed',
'NumberOfOpenCreditLinesAndLoans':'open credits',
'NumberOfTimes90DaysLate':'90 passed',
'NumberRealEstateLoansOrLines':'Real Estate credits',
'NumberOfTime60-89DaysPastDueNotWorse' : '60-89 passed',
'NumberOfDependents':'dependents'}

df_gmc = df_gmc.rename(columns=renamer)
```

```python
df_gmc.columns
```

```python
df_gmc = df_gmc[~df_gmc['dependents'].isnull()]
df_gmc = df_gmc[df_gmc['60-89 passed']!=98]
df_gmc = df_gmc[df_gmc['60-89 passed']!=96]
```
