```python
import kagglehub
import pandas as pd
import os
import glob

```

```python

path = kagglehub.dataset_download("kabure/german-credit-data-with-risk")

# Find the CSV file within the downloaded directory
csv_files = glob.glob(os.path.join(path, '*.csv'))

if csv_files:
    # Assuming the first CSV file found is the correct one
    data_file_path = csv_files[0]
    german_dataset_raw = pd.read_csv(data_file_path)
    print(german_dataset_raw.head())
else:
    print(f"No CSV files found in the downloaded dataset directory: {path}")
    german_dataset_raw = None

print(german_dataset_raw.head())
# There are no duplicates
german_df = german_dataset_raw.copy()


def columndropper(df,cols_to_drop):
  df = df.drop(columns=cols_to_drop)
  return df
german_df = columndropper(german_df,['Unnamed: 0'])
```
